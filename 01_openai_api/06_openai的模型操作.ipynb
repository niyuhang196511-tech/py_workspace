```python
from openai import OpenAI
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
base_url = os.getenv("BL_BASE_URL")
api_key = os.getenv("BL_API_KEY")
```

```python
client = OpenAI(base_url=base_url, api_key=api_key)
```

```python
# 获取模型列表
models = client.models.list()

models.data
```
